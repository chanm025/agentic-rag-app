## Facebook Prophet 
##### - open-source forecasting library 
##### - decomposes a time series into trend, seasonality, holidays and errors

### y(t) = g(t) + s(t) + h(t) + et
##### - Trend (g(t)) is for long-term growth or decline.
##### - Seasonality (s(t)) includes recurring patterns such as weekly, monthly, or yearly fluctuations.
##### - Holiday effects (h(t)) account for irregular but predictable events, such as Black Friday or national holidays.
##### - Error (et) is the residual noise not explained by the other components.

```python
#!pip install prophet
from prophet import Prophet
from sklearn.metrics import mean_absolute_percentage_error, mean_absolute_error, mean_squared_error
```

```python
#making a function of Prophet model to use for each currency 

def run_prophet_currency(
    df,
    currency_col,
    date_col='Time Serie',
    train_ratio=0.7,
    weekly=True,
    yearly=True,
    seasonality_mode='additive',
    plot=True
):

    #preparing data for Prophet
    #ds = datestamp, y = value to forecast
    df_ts = df[[date_col, currency_col]].copy()
    df_ts = df_ts.rename(columns={date_col: 'ds', currency_col: 'y'}) #have to change to ds and y (Prophet only allows these)
    
    # ensure valid dates
    df_ts['ds'] = pd.to_datetime(df_ts['ds'], format='%d-%m-%Y', errors='coerce')
    df_ts = df_ts.dropna(subset=['ds', 'y'])
    
    # split
    split = int(train_ratio * len(df_ts))
    train_df = df_ts.iloc[:split]
    test_df  = df_ts.iloc[split:]
    
    # fitting the Prophet model 
    #trains Prophet on data with both daily and yearly seasonal components
    proph = Prophet(
        daily_seasonality=False,
        weekly_seasonality=weekly,
        yearly_seasonality=yearly)
    proph.fit(train_df)
    
    # predicting on test time stamps 
    forecast = proph.predict(test_df[['ds']])
    
    # joining predictions with actual values (merging)
    #forecast contains columns such as yhat(predicted), yhat_lower, and yhat_upper
    merged = pd.merge(
        test_df,
        forecast[['ds', 'yhat']],
        on='ds', #specifies that merge should happen on the 'ds' column, must be present in both DataFrames
        how='inner' #specifies an inner join -> only rows with matching 'ds' values in both DataFrames will be included in the result
    )

    #computing error metrics 
    mae = mean_absolute_error(merged['y'], merged['yhat'])
    rmse = np.sqrt(mean_squared_error(merged['y'], merged['yhat']))
    mape = np.mean(np.abs((merged['y'] - merged['yhat']) / merged['y'])) * 100

    metrics = {
    'MAE': round(mae, 4),
    'RMSE': round(rmse, 4),
    'MAPE (%)': round(mape, 4)
    }

    #plotting actual vs predicted test set 
    plt.figure(figsize=(10,5))
    # training data
    plt.plot(train_df['ds'], train_df['y'], label='Train', color='black')
    # actual test values
    plt.plot(test_df['ds'], test_df['y'], label='Actual (Test)', color='green')
    # predicted test values
    plt.plot(merged['ds'], forecast['yhat'], label='Forecast (Test)', color='blue')
    plt.legend()
    plt.title(f"Prophet Forecast: {currency_col}")
    plt.xlabel("Date")
    plt.ylabel(currency_col)
    plt.show()
    
    #plot seasonal components
    proph.plot_components(forecast)
    plt.show()

    return {
    "model": proph,
    "train": train_df,
    "test": test_df,
    "predictions": merged,
    "metrics": metrics
}
```

```python
results = run_prophet_currency(df, "SRI LANKA - SRI LANKAN RUPEE/US$")

results["metrics"]
```
